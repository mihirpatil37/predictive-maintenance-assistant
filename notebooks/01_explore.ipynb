```python
import pandas as pd
import matplotlib.pyplot as plt

cols = ['unit', 'cycle', 'op1', 'op2', 'op3'] + [f"s{i}" for i in range(1,22)]
df = pd.read_csv("../data/raw/train_FD001.txt", sep=r"\s+", header=None, names=cols, index_col=False)

print(df.shape)
print(df.groupby("unit")["cycle"].max().describe())
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
for unit in [1, 2, 3, 4, 5]:
    d = df[df["unit"] == unit]
    ax.plot(d["cycle"], d["s11"], label=f"engine {unit}")
ax.set(xlabel="cycle", ylabel="sensor 11", title="Sensor 11 drifts as engines degrade")
ax.legend()
plt.show()

fig.savefig("../docs/sensor11_degradation.png", dpi=150, bbox_inches="tight")
```
